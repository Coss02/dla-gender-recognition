## 2. Plot loss curves

```python
def plot_metric_comparison(histories, metric_key, title, ylabel):
    """Plot a single metric across all models."""
    fig, ax = plt.subplots(figsize=(10, 5))
    for name, history in histories.items():
        epochs = range(1, len(history[metric_key]) + 1)
        ax.plot(epochs, history[metric_key], label=name, marker="o", markersize=3)
    ax.set_xlabel("Epoch")
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    ax.legend()
    plt.tight_layout()
    return fig


# Uncomment after loading histories:
# fig = plot_metric_comparison(histories, "train_loss", "Training Loss", "Loss")
# fig.savefig("../outputs/learning_curves_train_loss.png", dpi=150)
# plt.show()

# fig = plot_metric_comparison(histories, "val_loss", "Validation Loss", "Loss")
# fig.savefig("../outputs/learning_curves_val_loss.png", dpi=150)
# plt.show()

# fig = plot_metric_comparison(histories, "val_accuracy", "Validation Accuracy", "Accuracy")
# fig.savefig("../outputs/learning_curves_val_accuracy.png", dpi=150)
# plt.show()
```

## 3. Train vs Val loss per model (overfitting check)

```python
def plot_train_vs_val(history, model_name):
    """Plot train and val loss side by side for a single model."""
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    epochs = range(1, len(history["train_loss"]) + 1)

    axes[0].plot(epochs, history["train_loss"], label="Train", marker="o", markersize=3)
    axes[0].plot(epochs, history["val_loss"], label="Val", marker="s", markersize=3)
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].set_title(f"{model_name} — Loss")
    axes[0].legend()

    axes[1].plot(epochs, history["val_accuracy"], label="Val Accuracy", marker="o", markersize=3, color="green")
    axes[1].plot(epochs, history["val_f1"], label="Val F1", marker="s", markersize=3, color="orange")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Score")
    axes[1].set_title(f"{model_name} — Metrics")
    axes[1].legend()
    axes[1].set_ylim(0, 1.05)

    plt.tight_layout()
    return fig


# Uncomment after loading histories:
# for name, history in histories.items():
#     fig = plot_train_vs_val(history, name)
#     fig.savefig(f"../outputs/{name}_train_vs_val.png", dpi=150)
#     plt.show()
```
